```python
from pathlib import Path
import pandas as pd
```

```python
project_root = Path.cwd()
file_path = project_root / "data" / "Nova_Ops_Entry.xlsx"

print(file_path)
print(file_path.exists())
```

```python
excel_file = pd.ExcelFile(file_path)
print(excel_file.sheet_names)
```

```python
df_purchases = pd.read_excel(
    file_path,
    sheet_name="Purchases",
    header=4
)

df_purchases.head()
```

```python
print("Shape:", df_purchases.shape)

print("\nColumns:")
print(df_purchases.columns.tolist())

print("\nData types:")
print(df_purchases.dtypes)
```

```python
missing_values = df_purchases.isna().sum()

print(missing_values)
```

```python
# الاختبار الاول: هل توجد صفوف فارغة بالكامل؟
fully_empty_rows = df_purchases.isna().all(axis=1)

print("Fully empty rows: ", fully_empty_rows.sum())
print("Thier indexes: ", df_purchases.index[fully_empty_rows].tolist())
```

```python
# الاختبار الثاني: ما السجلات التي تفتقد رقم المستند؟
missing_doc_no = df_purchases["Doc_No"].isna()

print("Rows missing Doc_No: ", missing_doc_no.sum())
print("Indexes: ", df_purchases.index[missing_doc_no].tolist())
```

```python
df_purchases.loc[missing_doc_no].isna().sum()
```

```python
print("Date dtype: ", df_purchases["Doc_Date"].dtype)

print("\nFirst 10 date values:")
print(df_purchases["Doc_Date"].head(10).to_string(index=False))

print("\nMissing dates:", df_purchases["Doc_date"].isna().sum())
```
